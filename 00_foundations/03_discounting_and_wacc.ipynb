# Discounting and WACC

## 1. Intuition

Companies are financed by:
- Equity (shareholders)
- Debt (lenders)

Each requires a return → WACC is the **blended required return**

## 2. Mathematics



## 3. Implementation

```python
import plt


def cost_of_equity(rf, beta, mrp):
    return rf + beta * mrp

def wacc(E, D, re, rd, tax):
    total = E + D
    return (E/total)*re + (D/total)*rd*(1-tax)

rf = 0.015
beta = 0.65
mrp = 0.055

re = cost_of_equity(rf, beta, mrp)

E = 0.85
D = 0.15
rd = 0.03
tax = 0.21

wacc_value = wacc(E, D, re, rd, tax)

re, wacc_value
```

## 4. Visualization

```python
import numpy as np

wacc_range = np.linspace(0.04, 0.08, 50)

plt.plot(wacc_range, 1 / wacc_range)
plt.title("Valuation Sensitivity to WACC")
plt.xlabel("WACC")
plt.ylabel("Relative Value")
plt.grid()
plt.show()
```

## 5. Application
- Used as **discount rate in DCF**
- Represents:
  - risk
  - capital structure
  - market expectations

## 6. Reflection
- WACC is the most important parameter in valuation
- Small changes → large valuation swings
- Must be scenario-based, not fixed
